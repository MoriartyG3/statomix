```python
def get_surv_df(df, surv_pair, col_names:list = None):
    if col_names is None:
        surv_df = df[[surv_pair.time_profile.col_name, surv_pair.event_profile.col_name]].copy()
    
    else:
        surv_df = df[[surv_pair.time_profile.col_name, surv_pair.event_profile.col_name]+col_names].copy()
    surv_df = surv_df.rename(
        columns={
            surv_pair.time_profile.col_name: "time",
            surv_pair.event_profile.col_name: "event",
        }
    )
    surv_df = surv_df.dropna()
    surv_df["event"] = surv_df["event"].astype(float).astype(int)#.astype(bool)
    surv_df['event'] = surv_df['event'].astype(str).map({'0': False, '1': True})

    return surv_df
```
